```python
import numpy as np
import pandas as pd
arr=np.array([50, 40, 30, 20, 10])
print("\n Numpy array")
print("Array: ", arr)
```

```python
print("First element:",arr[0])
print("Second element:",arr[1])
print("Third element:",arr[2])
print("Fourth element:",arr[3])
print("First element:",arr[0])
print("Fifth element:",arr[4])
print("Last element:",arr[-1])
```

```python
print("Array Slicing\n")
print("first 4 elements:",arr[:4])
print("elements from index 3:",arr[3:])
print("last 3 elements:", arr[-3:])
```

```python
print("Basic array operations\n")
print("Add 100:", arr+100)
print("Subtract 50:", arr-50)
print("Multiply 2:", arr*2)
print("Divide 10:", arr/10)
```

```python
print("Mathematical operations\n")
print("Sum:",np.sum(arr))
print("Mean:",np.mean(arr))
print("Maximum:",np.max(arr))
print("Minimum:",np.min(arr))
```

```python
arr=np.array([10,20,30,40,50,60])
new_arr=arr.reshape(6, 1)
print("\nOriginal array:\n")
print(arr)
print("\nReshaped array:\n")
print(new_arr)
```

```python
data={
    "Name":["Amrutha", "Anagha", "Ankita", "Yashika"],
    "Marks":[90, 88, 100, 91],
    "Attendance":[100, 67, 75, 85]
}
df=pd.DataFrame(data)
print("\nPandas DataFame\n")
print(df)
```

```python
print("Name Column:")
print(df["Name"])
print("Marks Column:")
print(df["Marks"])
```

```python
print("Name and Marks Columns:")
print(df[["Name", "Marks"]])
```

```python
print("Average Marks:",df["Marks"].mean())
print("Highest Marks:",df["Marks"].max())
print("Lowest Marks:",df["Marks"].min())
```

```python
print("\nOriginal DataFrame\n")
print(df)
print("\nRow Indexing\n")
print("\nFirst Row:\n")
print(df.iloc[0])
print("\nSecond Row:\n")
print(df.iloc[1])
print("\nThird Row:\n")
print(df.iloc[2])
print("\nFourth Row:\n")
print(df.iloc[3])
```

```python
print("\nSpecific Values\n")
print("Marks of Amrutha:", df.iloc[0,1])
print("Marks of Anagha:", df.iloc[1,1])
print("Branch of Ankita:", df.iloc[2,2])
print("Last Name:", df.iloc[3,0])
```

```python
print("\nData Filtering\n")
result=df[df['Marks']>90]
print("Students with marks greater than 90:\n")
print(result)
```
